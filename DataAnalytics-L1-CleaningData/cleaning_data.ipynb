```python
# Rows missing Price: do they always have Quantity AND Total present?
missing_price = df[df['Price Per Unit'].isnull()]
missing_price[['Quantity', 'Total Spent']].isnull().sum()
```

```python
# Rows missing Quantity: do they always have Total also missing?
missing_qty = df[df['Quantity'].isnull()]
missing_qty[['Price Per Unit', 'Total Spent']].isnull().sum()
```

```python
df['Price Per Unit'] = df['Price Per Unit'].fillna(df['Total Spent'] / df['Quantity'])
```

```python
df['Price Per Unit'].isnull().sum()
```

```python
df = df.dropna(subset=['Quantity', 'Total Spent'])
df.shape
```

```python
df['Item'] = df['Item'].fillna('Unknown')
```

```python
df['Discount Applied'] = df['Discount Applied'].fillna(False)
```

```python
df['Transaction Date'] = pd.to_datetime(df['Transaction Date'])
df.info()
```

```python
df['Discount Applied'] = df['Discount Applied'].astype(bool)
df.dtypes
```

```python
df = df.reset_index(drop=True)
df.to_csv('cleaned_data.csv', index=False)
```
